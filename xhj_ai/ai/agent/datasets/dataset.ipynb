```python
# 魔搭专属的数据集加载工具
from modelscope.msdatasets import MsDataset

full_ms_ds = MsDataset.load(
    "DAMO_NLP/yf_dianping",  # 数据集ID
    subset_name="default",  # 该数据集只有一个default
    split="train"  # 数据集只包含训练集 100% 所有数据
)

full_ms_ds[0]
```

```python
# train validate test
# 魔搭社区的数据集，转成底层兼容Hugging Face Dataset
full_hf_ds = full_ms_ds.to_hf_dataset()
# 10% 的数据作为train, seed 随机采样 
split1 = full_hf_ds.train_test_split(test_size=0.1, seed=42)
train_temp_hf= split1["train"]
test_hf = split1["test"]

split2 = train_temp_hf.train_test_split(test_size=0.1,seed=42)
train_hf = split2["train"]
val_hf = split2["test"]

print(f"训练集train: {len(train_hf)}")
print(f"验证集val: {len(val_hf)}")
print(f"测试集val: {len(test_hf)}")
print("\n 单条样本：",train_hf[0])
```
